```python
class euron_course:
  def __init__(self, course_name, course_mentor,course_price):
    self.course_name = course_name,
    self.course_mentor = course_mentor,
    self.course_price = course_price

  def display_course(self):
    print(f"Course Name: {self.course_name}, Course Mentor: {self.course_mentor}, Course Price: {self.course_price}")

  def update_price(self,new_price):
    self.course_price = new_price
    print(f"Course Price is updated to:",{self.course_price})

  def enroll_student(self, student_name):
    print(f"{student_name} is enrolled to {self.course_name}")
```

```python
class euron_course:
  def __init__(self, course_name, course_mentor,course_price):
    self.course_name = course_name,
    self.course_mentor = course_mentor,
    self.__course_price = course_price

  def display_course(self):
    print(f"Course Name: {self.course_name}, Course Mentor: {self.course_mentor}, Course Price: {self.course_price}")

  def update_price(self,new_price):
    self.course_price = new_price
    print(f"Course Price is updated to:",{self.course_price})

  def enroll_student(self, student_name):
    print(f"{student_name} is enrolled to {self.course_name}")
#Encapsulation --> You are restricting an access of certain entities inside a class.
  def get_price(self): # encapsulation
    return self.__course_price
```

```python
#Inheritence --> If we are taking some property from our parents is called inheritance
class euron:
  def __init__(self,course_name,course_mentor,course_price):
    self.course_name = course_name
    self.course_mentor = course_mentor
    self.course_price = course_price

class euron_course(euron):
  def __init__(self,course_name,course_mentor,course_price,course_duration):
    super().__init__(course_name,course_mentor,course_price)
    self.course_duration = course_duration
  def display_course(self):
    print(f"Course Name: {self.course_name}, Course Mentor: {self.course_mentor}, Course Price: {self.course_price}, Course Duration: {self.course_duration}")
```
