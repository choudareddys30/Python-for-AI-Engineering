```python
#recursive function --> running the function again and again
def fact(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n*fact(n-1)
```

```python
fact(5)
```

```python
def string_rev(s):
  if len(s) == 0:
    return ""
  else:
    return s[-1] + string_rev(s[:-1])
```
