```python
l = ["Chinna",1,2,3.4,True,"Hello","Everyone",False,3+4j]
```

```python
string_list=[]
int_list=[]
float_list=[]
bool_list=[]
complex_list=[]
for i in l:
  if type(i) == str:
    string_list.append(i)
  elif type(i) == int:
    int_list.append(i)
  elif type(i) == float:
    float_list.append(i)
  elif type(i) == bool:
    bool_list.append(i)
  else:
    complex_list.append(i)
print(string_list)
print(int_list)
print(float_list)
print(bool_list)
print(complex_list)
```

```python
l = [1,2,3,4,5,10.4,20.4,"Hello","Hi",True, False, 10+4j, 5+1j]
```

```python
int_list = []
float_list = []
string_list = []
bool_list = []
complex_list = []

for i in l:
  if type(i) == int:
    int_list.append(i)
  elif type(i) == float:
    float_list.append(i)
  elif type(i) == str:
    string_list.append(i)
  elif type(i) == bool:
    bool_list.append(i)
  else:
    complex_list.append(i)
print("int_list",int_list)
print("float_list",float_list)
print("string_list",string_list)
print("bool_list",bool_list)
print("complex_list",complex_list)
```
