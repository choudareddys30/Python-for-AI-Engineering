```python
# we can mention any number of try blocks and except blocks, try-except

try: # identify the error
  a = int(input("Enter an Integer a:"))
  b = int(input("Enter an Integer b:"))
  c=a/b
  print(c)
except Exception as e: # handle those errors 
  print(e)
finally: # It is not a mandatory block
  print("Hello")
```

```python
# we can mention any number of try blocks and except blocks, try-except

try: # identify the error
  a = int(input("Enter an Integer a:"))
  b = int(input("Enter an Integer b:"))
  c=a/b
  print(c)
except Exception as e: # handle those errors 
  print(e)
finally: # It is not a mandatory block
  try:
    d = int(input())
  except:
    pass
```

```python
try: 
  print("In this task enter a number, check for even, if even then add 10 for it")
  number = int(input("Enter a Number:"))
  try: 
    if number % 2 ==0:
      print(number," is even")
      try:
        print("As number is even add 10 to the final output",number+10)
      except Exception as e:
        print(e)
    else:
      print(number, "is odd")
  except Exception as e:
    print(e)
except Exception as e:
  print(e)
```

```python
try:
  a=int(input())
  b=int(input())
  print("a=",a)
  print("b=",b)
  print(a/b)
except ZeroDivisionError:
  print('Cannot divide by zero')
except ValueError:
  print("Something went wrong")
else:
  print("There is no error in the code")
finally:
  print("This is a finally Block")
```

```python
#raise


age = 18

if age < 18:
  raise ValueError("Age Must be 18 or above")
print("Eligible")
```

```python
class InvalidAgeError(Exception):
  pass

age = 21

try:
  if age < 18:
    raise InvalidAgeError("Not Eligible")
except InvalidAgeError as e:
  print(e)
```
