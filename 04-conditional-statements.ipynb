```python
userid = input("Enter User Name")
password = input("Enter Password")

print("UserId:",userid)
print("Password:",password)

if userid == 'Python' and password == 'Pass@123':
  print("It is a valid user, allowed to login")
elif userid != 'Python' and password != 'Pass@123':
  print("Userid and Password are incorrect")
elif userid != 'Python':
  print("UserID is incorrect")
else:
  print("Password is incorrect")
```

```python
# Checking even or odd number

number = int(input("Enter your number"))

if number % 2 == 0:
  print(number, " is an even number")
else:
  print(number," is an odd number")
```

```python
#nested if condition

basic_course = input("Enter the Course Name:")
advanced_course = input("Enter the Course Name")

print("Basic Course:",basic_course)
print("Advanced Course:",advanced_course)

if basic_course == "Python":
  if advanced_course == "Data science":
    print("You will become a data scientist")
  elif advanced_course == "big data":
    print("You will become a Data Engineer")
  elif advanced_course == "data analytics":
    print("You will become a Data Analyst")
  elif advanced_course == "Gen AI":
    print("You will become a GEN AI Engineer")
  else:
    print("You are a Python Dev")
else:
  if basic_course == "js":
    if advanced_course == "node js":
      print("You are MERN Stack Dev")
    else:
      print("You are a JS Dev")
```
